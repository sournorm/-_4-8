## Завдання 2. Скалярний добуток і довжина векторів

Дано вектори:
```python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])
```

1. Обчисліть скалярний добуток двома способами: `np.dot()` і оператором `@`.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами.

````python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])

dot_np = np.dot(u, v)
dot_op = u @ v
print("Скалярний добуток (np.dot):", dot_np)
print("Скалярний добуток (@):", dot_op)

norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
print("Норма u:", norm_u)
print("Норма v:", norm_v)

cos_theta = dot_np / (norm_u * norm_v)
print("Косинус кута:", cos_theta)
````

## Завдання 15. Комутатор двох матриць

Для кожної пари матриць обчисліть комутатор
\[
[A,B] = AB - BA.
\]

Перша пара:
```python
A1 = np.array([[3, 0],
               [0, 7]])
B1 = np.array([[2, 0],
               [0, 5]])
```

Друга пара:
```python
A2 = np.array([[0, 2],
               [-1, 0]])
B2 = np.array([[4, 0],
               [0, 1]])
```

Визначте, для якої пари виконується `AB = BA`. Для перевірки використайте `np.allclose()`.

````python
A1 = np.array([[3, 0], [0, 7]])
B1 = np.array([[2, 0], [0, 5]])
A2 = np.array([[0, 2], [-1, 0]])
B2 = np.array([[4, 0], [0, 1]])

comm1 = (A1 @ B1) - (B1 @ A1)
comm2 = (A2 @ B2) - (B2 @ A2)

print("Комутатор 1:\n", comm1)
print("Комутатор 2:\n", comm2)

print("AB = BA (пара 1):", np.allclose(A1 @ B1, B1 @ A1))
print("AB = BA (пара 2):", np.allclose(A2 @ B2, B2 @ A2))
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

X = np.vstack([np.ones(len(x)), x]).T
coeffs, residuals, _, _ = np.linalg.lstsq(X, y, rcond=None)

a, b = coeffs
print(f"a = {a:.4f}, b = {b:.4f}")

y_pred = a + b * x
print("Прогнозні y:", y_pred)

rss = residuals[0] if len(residuals) > 0 else np.sum((y - y_pred)**2)
print("Сума квадратів залишків:", rss)
````
